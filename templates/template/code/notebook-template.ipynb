## Problem

Write a function that returns the *n*-th Fibonacci number (0-indexed), then print the first ten values.

````python
def fibonacci(n):
    """Return the n-th Fibonacci number (0-indexed)."""
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a


# A short comment, to show off comment styling too
values = [fibonacci(n) for n in range(10)]
print("First 10 Fibonacci numbers:", values)
````
